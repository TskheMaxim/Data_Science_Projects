**2) What is a Series?**
<br />
-> A Pandas Series is like a column in a table.

```python
a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar)
```

a) Labels
<br />
-> If nothing else is specified, the values are labeled with their index number. First value has index 0, second value has index 1 etc.
<br />
-> This label can be used to access a specified value.

```python
print(myvar[1])
```

b) Create Labels
<br />
-> With the 'index' argument, you can name your own labels.

```python
myvar = pd.Series(a, index = ["a", "b", "c"])

print(myvar)
```

-> When you have created labels, you can access an item by referring to the label.

```python
print(myvar["a"])
```

c) Key/Value Objects as Series
<br />
-> You can also use a key/value object, like a dictionary, when creating a Series.

```python
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

-> Specify only the items you want to include in the Series.

```python
myvar = pd.Series(calories, index = ["day2"])

print(myvar)
```

**3) DataFrames**
<br />
-> Data that is more than 1 column is multi-dimensional tables, called DataFrames.
<br />
-> Series is like a column, a DataFrame is the multi-dimensional table.

```python
#COmbination fo dict and list
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data)

print(df)
```

a) Locate Row
<br />
-> Pandas use the loc attribute to return one or more specified row(s)

```python
#refer to the row index:
print(df.loc[0])
```

```python
#use a list of indexes:
print(df.loc[[0, 2]])
```

b) Named Indexes
<br />
-> With the index argument, you can name your own indexes.

```python
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print (df)
```

c) Locate Named Indexes
<br />
->Use the named index in the loc attribute to return the specified row(s).

```python
print(df.loc["day1"])
```
