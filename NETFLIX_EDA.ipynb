# Dummy Splitting Genres

```python
import pandas as pd

# 1. Define your verification question column name
col_genres = 'genres'

# Check if the column exists so it's safe to run multiple times
if col_genres in df.columns:
    # 2. YOUR ORIGINAL CODE (Kept exactly as it is)
    purpose_dummies = df[col_genres].str.get_dummies(sep=', ')
    df = pd.concat([df, purpose_dummies], axis=1).drop(columns=[col_genres])
    
    # 3. THE STRUCTURAL FIX: Rename them so they belong to the question
    # This locks them under the parent question header visually
    rename_dict = {c: f"{col_genres} ➔ {c}" for c in purpose_dummies.columns}
    df = df.rename(columns=rename_dict)

# ==========================================
# DISPLAY THE GROUPED CHOICES TOGETHER
# ==========================================
# This automatically fetches all columns starting with your verification question prefix
grouped_columns = [c for c in df.columns if c.startswith(col_genres)]
df[grouped_columns].head(10)
```
